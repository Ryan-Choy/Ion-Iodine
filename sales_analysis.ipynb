# Daily sales analysis
This example uses five invented daily sales values.
We compare the average with the median.

```python
from statistics import mean, median
sales = [120, 180, 150, 250, 300]
print(f"Days: {len(sales)}")
print(f"Average daily sales: ${mean(sales):.2f}")
print(f"Median daily sales: ${median(sales):.2f}")
```

The average is \\$200.00 and the median is \\$180.00. Higher
sales values pull the average above the middle observation. 

```python
print(f"Lowest daily sales: ${min(sales):.2f}")
print(f"Highest daily sales: ${max(sales):.2f}")
```

The lowest daily sale is \\$120.00 and the highest daily sale is \\$300.00.
